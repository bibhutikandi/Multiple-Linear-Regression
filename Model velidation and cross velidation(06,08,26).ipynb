```python
features=df[['Daily Time Spent on Site','Age','Area Income','Daily Internet Usage','Male']]
features.head()
```

```python
target=df[['Clicked on Ad']]
```

```python
features.boxplot()
```

```python
from sklearn.feature_selection import f_classif

f_clas= f_classif(features,target)

pd.Series(f_clas[0],index=features.columns).sort_values(ascending=False)
```

```python
features.drop(columns=['Male'],inplace=True)
```

```python
features.head()
```

##Hold Out Method

```python
###Train_Test_split
x_train,x_test,y_train,y_test=train_test_split(features,target,train_size=0.8,random_state=100,stratify=df['Clicked on Ad'])

print(x_train.shape)

print(x_test.shape)

print(y_train.shape)

print(y_test.shape)
```

```python
std_sca= StandardScaler()

x_train= pd.DataFrame(std_sca.fit_transform(x_train),columns=x_train.columns)

x_test= pd.DataFrame(std_sca.transform(x_test),columns=x_test.columns)
```

```python
x_train.head()
```

```python
log_model= LogisticRegression()

log_model.fit(x_train,y_train)

#### Training Validation

y_pred= log_model.predict(x_train)

train_acc= accuracy_score(y_train,y_pred)

### Testing validation

y_pred1= log_model.predict(x_test)

test_acc= accuracy_score(y_test,y_pred1)

print('Training_Acc:',train_acc,',','Testing_Acc:',test_acc)
```

###Suffle Split

```python
shuf_split=ShuffleSplit(n_splits=10,test_size=0.2,random_state=50)

for train_split, test_split in shuf_split.split(features,target):

    print(train_split,test_split)
```

```python
cross=cross_validate(LogisticRegression(),features,target,scoring='accuracy',cv=shuf_split,return_train_score=True)
cross
```

```python
cross= cross_validate(LogisticRegression(),features,target,scoring='accuracy',cv=shuf_split,return_train_score=True)

test_acc=cross['test_score'].mean().round(2)

train_acc= cross['train_score'].mean().round(2)

print('Training_Acc:',train_acc,',','Testing_Acc:',test_acc)
```

KFold
